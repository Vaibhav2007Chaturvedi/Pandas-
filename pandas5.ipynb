**Merging 2 dataframes:**

```python
import pandas as pd
import numpy as np
```

```python
employees=pd.DataFrame({
    'employee_id':[1,2,3,4,5],
    'name':['John','Anna','Peter','Linda','Bob'],
    'department':['HR','IT','Finance','IT','HR']
})
#DataFrame 2: Salary Information:
salaries=pd.DataFrame({
    'employee_id':[1,2,3,6,7],
    'salary':[60000,80000,65000,70000,90000],
    'bonus':[5000,10000,7000,8000,12000]
})
print(employees)
print(salaries)
print(pd.merge(employees,salaries))
print(pd.merge(employees,salaries,on="employee_id"))
```

```python
employees=pd.DataFrame({
    'employee_id':[1,2,3,4,5],
    'name':['John','Anna','Peter','Linda','Bob'],
    'department':['HR','IT','Finance','IT','HR']
})
#DataFrame 2: Salary Information:
salaries=pd.DataFrame({
    'employee_id':[1,2,3,6,7],
    'salary':[60000,80000,65000,70000,90000],
    'bonus':[5000,10000,7000,8000,12000]
})
merged=pd.merge(
    employees,
    salaries,
    on="employee_id",
    how='inner'
)
print(merged)

#Inner Merge:It will gives only matching records from both DataFrame.
```

```python
employees=pd.DataFrame({
    'employee_id':[1,2,3,4,5],
    'name':['John','Anna','Peter','Linda','Bob'],
    'department':['HR','IT','Finance','IT','HR']
})
#DataFrame 2: Salary Information:
salaries=pd.DataFrame({
    'employee_id':[1,2,3,6,7],
    'salary':[60000,80000,65000,70000,90000],
    'bonus':[5000,10000,7000,8000,12000]
})
merged=pd.merge(
    employees,
    salaries,
    on='employee_id',
    how='left'
)
print(merged)
#it will keeps all records from the dataFrame (employees)
# as Here IDs 4 and 5 don't have salary information, so we get: NaN
```

**Concatenation of Two DataFrames in Pandas:**

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})
df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
#Concatenate Row-wise
result=pd.concat([df1,df2])
print(result)
result=pd.concat([df2,df1]) #By default axis=0 .
print(result)
#Concatenate Column-wise:
result=pd.concat([df1,df2],axis=1)
print(result)
```

```python
f1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})
df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
result=pd.concat([df1,df2],ignore_index=True)
print(result)
#Reset Index: to get a continous index.
```

**Joining 2 Data Frame**

```python
df1=pd.DataFrame({
    'name':['Alice','Bob','Charlie']
},
index=[1,2,3]
)
df2=pd.DataFrame({
    'score':[85,90,75]
},
index=[2,3,4]
)
print(df1)
print(df2)
a=df1.join(df2)
print(a)
a=df2.join(df1)
print(a)
a=df1.join(df2,how='outer')
print(a)
```
