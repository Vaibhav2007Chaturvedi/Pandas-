PANDAS DATAFRAME:  (PART1)
WHAT IS DATAFRAME: A dataframe is a two-dimensional labeled data structure in Pandas.
It looks like a table with:
Rows
Columns
Index
Values

```python
#Creating DataFrame from a dictionary :
import pandas as pd 
data={
    'Name':['Vaibhav','Harsh','Parth'],
    'Age':[20,21,24],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
print(s)
```

```python
#Creating a dataframe from a list of lists:
data=[
    ['Vaibhav',20,85],
    ['Hrash',24,98],
    ['Chitransh',21,99]
]
s=pd.DataFrame(data)
print(s)
#By default, columns are numbered 0, 1, 2....
#Giving a column Name:
col=['Name','Age','Marks']
s=pd.DataFrame(data,columns=col)
print(s)
```

```python
#Creating a DataFrame from Numpy Array:
import numpy as np
arr=np.array([
    [10,20],
    [30,40],
    [50,60]
])
s=pd.DataFrame(arr)
print(s)
#..............................
#Giving Custom Index: we can provide custom row labels using index.
data = {
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Marks': [85, 90, 88]
}
ind=['Student1','Student2','Student3']
s=pd.DataFrame(data,index=ind)
print(s)
```

Selection and  Indexing of Columns:
Selection means selecting a particular column or multiple columns from a DataFrame.

```python
#1.Selecting a single columns:
import pandas as pd
data={
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
print(s['Name'])
print(s['Marks'])
print(s[['Name','Marks']])
#..........................
#2.Selecting a Columns Using Dot Notation:
print(s.Name)
# NOTE:
# ⚠️ Limitation

# 1.Dot notation does not work reliably when:

# 2.Column name contains spaces
# 3.Column name has special characters
# 4.Column name conflicts with a DataFrame method
```

```python
#3.Columns Indexing using .iloc --->.iloc selects data based on integer position.
data={
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
print(s.iloc[:,0]) #Select first column
print(s.iloc[:,1]) #Select second column
print(s.iloc[:,[0,2]]) #Selecting Multiple Columns using .iloc

```

Creating New Columns:
A new column can be added to a DataFrame by assigning values to a new column name.


```python
import pandas as pd
data = {
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
s['City']=['Delhi','Noida','Lucknow']
print(s)
#NOTE: The number of values should match the number of rows.
#............................
#Creating a Column with a Single Value
s['Country']='India'
print(s)
#..............................
#Creating a Column from Existing Columns
s['Marks_10']=s['Marks']/10
print(s)
#...............................
#Creating a Column using a Condition
s['Results']=s['Marks']>=40
print(s)
```

Removing Columns:
We can remove/delete columns from a DataFrame using different methods.

```python
#  1.Using drop():The drop() method is commonly used to remove columns.
import pandas as pd
data = {
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
# s=s.drop('Age',axis=0) #It will give error:
#axis=1 → means column
# axis=0 → means row
s=s.drop('Age',axis=1)
print(s)
#.....................
#Removing multiple data:We can remove multiple columns by passing a list.
data = {
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
s=s.drop(['Age','Marks'],axis=1)
print(s)
#..........................
#Using Columns Parameter:instead of axis=1, we can directly use columns=.
data = {
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
s=s.drop(columns=['Age'])
print(s)
#..............................
#Using del:We can also remove a column using del.
data = {
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
del s['Age']
print(s)
#NOTE:del directly removes the column from the DataFrame.
#.......................
#Using pop():
# pop() removes a column and returns the removed column as a Series.
data = {
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
age=s.pop('Age')
print(age)
print(s)
#NOTE:The Age column is removed from df, and its data is stored in age.
```

```python
# inplace :The inplace parameter is used with methods like drop() to decide whether the original 
# DatFrame should be modified .
import pandas as pd
data = {
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
s.drop('Age',axis=1,inplace=True)
print(s)
```

**** Important Difference
inplace=False — Default
df.drop('Age', axis=1)

--> Original df does not change.

inplace=True
df.drop('Age', axis=1, inplace=True)

--> Original df changes directly.

--->Remember

inplace=True → Make the change directly in the original DataFrame.
inplace=False → Return a modified copy and keep the original unchanged.
