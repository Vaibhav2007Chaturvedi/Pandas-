Pandas Series:A Series is a one-dimensional labeled array capable of holding any data type.The axis labels are collectively called the index.

Creating Series:There are multiple ways to create a pandas Series:

```python
#Creating Series from a list:
import pandas as pd
data=[10,20,30,40,50]
s=pd.Series(data)
print(s)
#............
#Creating Series with Custom Index:
#We can give our own labels/index while creating a Series.
data=[10,20,30,40,50]
labels=['a','b','c','d','e']
s=pd.Series(data,index=labels)
print(s)
# NOTE:The numbers of labels must be equal to the number of values.
#...................
#Changing the index after creating Series.
data=[10,20,30,40,50]
s=pd.Series(data)
labels=['a','b','c','d','e']
s.index=labels
print(s)
```

```python
#Creating Series from Numpy Array:
import numpy as np
import pandas as pd
arr=np.array([10,20,30,40])
s=pd.Series(arr)
print(s)
labels=['a','b','c','d']
s.index=labels
print(s)
```

```python
#Creating a series from dictionary:
data={
    'Maths':85,
    'Python':54,
    'Dsa':78,
    'C++':90
}
s=pd.Series(data)
print(s)
```
