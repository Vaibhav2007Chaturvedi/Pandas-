Selecting Rows:
Row selection means selecting one or more rows from a DataFrame.

We mainly use:

1: .loc[] → selection using labels/index
2: .iloc[] → selection using integer position

```python
#1.Selecting a Single rows using .loc: it select rows based on their index label.
import pandas as pd
data={
    'Name':['Vaibhav','Harsh','Parth'],
    'Age':[20,21,24],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
print(s)
print(s.loc[0])
print(s.loc[1])
print(s.loc[[0,2]])
print(s.loc[[1,2]])
```

```python
#2.Selecting Rows using .iloc[]:it selects rows based on their integer position .
import pandas as pd
data={
    'Name':['Vaibhav','Harsh','Parth'],
    'Age':[20,21,24],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
print(s.iloc[0])
print(s.iloc[1])
print(s.iloc[[0,1]])
print(s.iloc[0:2])
print(s.iloc[0:3])
```

```python
#3.Selecting rows and columns together:We can select  both rows and columns.
data={
    'Name':['Vaibhav','Harsh','Parth'],
    'Age':[20,21,24],
    'Marks': [85, 90, 88]
}
s=pd.DataFrame(data)
print(s)
print(s.loc[0:1,['Name','Marks']])
print(s.loc[0:2,['Name','Age']])
print(s.iloc[0:2,0:2])
#NOTE:First 2 rows and first 2 columns.
```

Selecting Subset of Rows and Columns:
A subset means selecting only a specific part of the DataFrame — particular rows and columns.

We mainly use:

.loc[] → select using labels
.iloc[] → select using integer positions

```python
import pandas as pd 
data = {
    'Name': ['Aman', 'Rahul', 'Priya', 'Riya'],
    'Age': [20, 21, 19, 22],
    'Marks': [85, 90, 88, 92]
}
s=pd.DataFrame(data)
print(s)
print(s.loc[0:2],['Name','Age'])
#Using .iloc[]:
print(s.iloc[0:3,[0,2]])
#Selecting a single row and multiple columns:
print(s.loc[1,['Name','Marks']])
print(s.loc[0:2, 'Name'])
```

Condition Selection:Condition selection means selecting rows from a DataFrame based on a condition.

It is also called Boolean Indexing.

```python
import pandas as pd
data = {
    'Name': ['Aman', 'Rahul', 'Priya', 'Riya'],
    'Age': [20, 21, 19, 22],
    'Marks': [85, 90, 88, 35]
}
s=pd.DataFrame(data)
print(s)

print(s[s['Marks']>80])
print(s[(s['Age']>19) & (s['Marks']>80)])
```
