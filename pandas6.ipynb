**GROUP BY AGGREGATION**

```python
import pandas as pd 
import numpy as np
```

**GroupBy**

```python
data={
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['S1','S1','S2','S2','S1','S2','S2','S1'],
    'Saales':[100,200,150,250,120,180,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01',periods=8)
}
df=pd.DataFrame(data)
print(df)
print(df.groupby('Category')) #GroupBy a Single Column
print(df.groupby('Store'))
a=df.groupby('Category')
print(a)
for i,v in a:
    print(i)
    print(v)
a=df.groupby('Store')
print(a)
for i,v in a:
    print(i)
    print(v)
```

```python
#GroupBy Multiple Columns:  We can group by more than one column.
data={
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['S1','S1','S2','S2','S1','S2','S2','S1'],
    'Saales':[100,200,150,250,120,180,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01',periods=8)
}
df=pd.DataFrame(data)
a=df.groupby(['Category','Store'])['Saales'].sum()
print(a)
a=df.groupby(['Category','Store'])['Quantity'].sum()
print(a)
```

```python
#GroupBy Multiple Aggregations: We can apply multiple aggregation functioon at once using agg().
data={
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['S1','S1','S2','S2','S1','S2','S2','S1'],
    'Saales':[100,200,150,250,120,180,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01',periods=8)
}
df=pd.DataFrame(data)
result=df.groupby('Category')['Saales'].agg(
    ['sum','mean','min','max']
)
print(result)
```

```python
#Aggregation on Multiple Colummns: We Can Calculate differnt aggregations for different columns.
data={
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['S1','S1','S2','S2','S1','S2','S2','S1'],
    'Saales':[100,200,150,250,120,180,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01',periods=8)
}
df=pd.DataFrame(data)
result=df.groupby('Category').agg({
    'Saales':'sum',
    'Quantity':'mean'
})
print(result)
```

```python
# Named Aggregation: A cleaner way to give meaningful names to the results:
data={
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['S1','S1','S2','S2','S1','S2','S2','S1'],
    'Saales':[100,200,150,250,120,180,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01',periods=8)
}
df=pd.DataFrame(data)
result=df.groupby('Category').agg(
    Total_Sales=('Saales','sum'),
    Average_Quantity=('Quantity','mean'),
    Maximum_Sales=('Saales','max')
)
print(result)
```

```python
#GroupBy+ reset_index(): groupby() often makes the grouping column an index.
ata={
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['S1','S1','S2','S2','S1','S2','S2','S1'],
    'Saales':[100,200,150,250,120,180,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01',periods=8)
}
df=pd.DataFrame(data)
result=df.groupby('Category')['Saales'].sum().reset_index()
print(result)
```
